### Subtask:
Sum the values from the 'Equity2001' to 'Equity2021' columns for each brand to create a new 'Total_Equity' column. This will represent the cumulative equity over the specified period.


**Reasoning**:
The subtask requires summing specific 'Equity' columns to create a new 'Total_Equity' column. First, I need to identify the columns that match the 'EquityYYYY' pattern for years 2001 to 2021 and then sum them row-wise.



````python
equity_columns = [f'Equity{year}' for year in range(2001, 2022) if f'Equity{year}' in df.columns]
df['Total_Equity'] = df[equity_columns].sum(axis=1)
print(df[['Brand_Name', 'Total_Equity']].head())
# I could tell here there was something wrong with the data but I wanted to see what the code would do
````

## Aggregate Equity by Country

### Subtask:

**Reasoning**:
To aggregate the total equity by country, I will group the DataFrame by the 'Country' column and then sum the 'Total_Equity' for each group.



````python
equity_by_country = df.groupby('Country')['Total_Equity'].sum()
print(equity_by_country.head())
# This is all very familiar from the Intro to Coding course and I completely understand it but couldn't have done it myself from memory.
````
